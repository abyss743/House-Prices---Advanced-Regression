###Section 1: Importing libraries and loading the data

```python
# importing necesssary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
# loading the training data
df = pd.read_csv('http://d2l-data.s3-accelerate.amazonaws.com/kaggle_house_pred_train.csv')

print("shape of dataset:",df.shape)
print("\n First 5 rows: ")
df.head()
```

### Section 2: Basic Dataset Info

```python
# Basic information on the dataset
print("*** Dataset Info ***")
df.info()

print("\n*** Statistical summary of numerical columns ***")
df.describe()
```

###Section 3: Target Variable Analysis (SalePrice)

```python
# SalePrice statistics
print("\n*** SalePrice stats ***")
print(df['SalePrice'].describe())
print("\n")

# ploting the distribution of SalePrice
plt.figure(figsize=(10,6))
sns.histplot(df['SalePrice'], kde=True, bins=50)
plt.xlabel('SalePrice')
plt.ylabel('Count')
plt.title('Distribution of SalePrice')
plt.show()
```

###Section 4: Missing Values Analysis

```python
# checking the missing values
missing_values = df.isnull().sum()
missing_values = missing_values[missing_values > 0].sort_values(ascending=False)
print("***Colums with missing values***")
print(missing_values)
print("\n Percentage of missing values")
print((missing_values/len(df)*100).round(2))
```

###Section 5: Correlation Analysis


```python
# selecting numeric columns
numeric_df = df.select_dtypes(include=[np.number])
# calculating correlation with SalePrice
correlation = numeric_df.corr()['SalePrice'].sort_values(ascending=False)
print("***Correlation with SalePrice***")
print(correlation.head(15))
```

###Section 6: Correlation Heatmap

```python
# assigning top 10 features most correlated with saleprice
top_corr_features = correlation.head(11).index
print("***Top 10 features most correlated with saleprice***")
# creating correlation matrix of these top features
top_corr_matrix = numeric_df[top_corr_features].corr()
# ploting heatmap
plt.figure(figsize=(12,8))
sns.heatmap(top_corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Matrix of Top 10 Features')
plt.show()
```

##Section 7: Key Feature Relationships (Scatter Plots)

```python
# overallqual vs saleprice
plt.figure(figsize=(10,6))
sns.scatterplot(x='OverallQual', y='SalePrice', data=df)
plt.title('Scatter Plot: OverallQual vs SalePrice')
plt.show()
# grlivarea vs saleprice
plt.figure(figsize=(10,6))
sns.scatterplot(x='GrLivArea', y='SalePrice', data=df)
plt.title('Scatter Plot: GrLivArea vs SalePrice')
plt.show()
```

###Section 8: Handling Missing Values

```python
# making the copy of the data so we don't change the original
df_clean = df.copy()
# columns where missing should be replaced with "none" data type
none_columns = ['PoolQC', 'MiscFeature', 'Alley', 'Fence', 'FireplaceQu', 'GarageType', 'GarageFinish', 'GarageQual', 'GarageCond', 'BsmtQual', 'BsmtCond', 'BsmtFinType1','BsmtFinType2','MasVnrType']
for col in none_columns:
    df_clean[col]=df_clean[col].fillna('None')
# filling the numeric columns with 0 or median
df_clean['LotFrontage'] = df_clean['LotFrontage'].fillna(df_clean['LotFrontage'].median())
df_clean['MasVnrArea'] = df_clean['MasVnrArea'].fillna(0)
df_clean['GarageYrBlt'] = df_clean['GarageYrBlt'].fillna(0)
# filling the remaining categorical columns with mode(most frequent value)
for col in df_clean.select_dtypes(include='object').columns:
    df_clean[col] = df_clean[col].fillna(df_clean[col].mode()[0])
# checking if there are still missing values
print("***Colums with missing values***")
print(df_clean.isnull().sum()[df_clean.isnull().sum()>0])
```

###Section 9: Log Transformation of SalePrice

```python
# creating log-transformed saleprice
df_clean['SalePrice_log'] = np.log1p(df_clean['SalePrice'])
# plotting the new distribution
plt.figure(figsize=(10,6))
sns.histplot(df_clean['SalePrice_log'], kde=True, bins=50)
plt.xlabel('SalePrice_log')
plt.ylabel('Count')
plt.title('Distribution of SalePrice_log')
plt.show()
```

###Section 10: Preparing Features & Target (Baseline)

```python
# separating the features and target
x = df_clean.drop(['Id','SalePrice', 'SalePrice_log'], axis=1)
y = df_clean['SalePrice_log']
print("features shape: ",x.shape)
print("target shape: ",y.shape)
print("\nfirst 5 feature colums")
print(x.columns[:5].tolist())
```

###Section 11: One-Hot Encoding

```python
# converting categorical columns into numbers using one-hot encoding
x_encoded = pd.get_dummies(x)
print("shape before encoding the features: ",x.shape)
print("shape after encoding the features: ",x_encoded.shape)
print("\n first 10 feature colums")
print(x_encoded.columns[:10].tolist())
```

###Section 12: Train-Validation Split

```python
from sklearn.model_selection import train_test_split
# splitting 80% of the data to train and 20% of the data for validation
x_train, x_val, y_train, y_val = train_test_split(x_encoded, y, test_size=0.2, random_state=42)
print("Training set size: ",x_train.shape)
print("Validation set size: ",x_val.shape)
```

###Section 13: Linear Regression (Baseline Model)

```python
# using linear regresssion model
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
# training the model
model = LinearRegression()
model.fit(x_train, y_train)
# making predictions on the validation set
y_pred = model.predict(x_val)
# calculating the mean squared error
rmse = np.sqrt(mean_squared_error(y_val, y_pred))
print("Validation RMSE score: ",round(rmse,5))
```

### Section 14: Random Forest Model

```python
# using the random forest model
from sklearn.ensemble import RandomForestRegressor
# Creating and training the model
rf_model = RandomForestRegressor(n_estimators=100, random_state=42,n_jobs = -1)
rf_model.fit(x_train, y_train)
# predict
y_pred_rf = rf_model.predict(x_val)
# calculating the mean squared error
rmse_rf = np.sqrt(mean_squared_error(y_val, y_pred_rf))
print("Validation RMSE score: ",round(rmse_rf,5))
```

###Section 15: XGBoost (Initial Version)

```python
# using xgbregressor model
from xgboost import XGBRegressor
# creating and trainning XGBoost model
xgb_model = XGBRegressor(n_estimators=1000, learning_rate=0.05, n_jobs = -1,max_depth=4,random_state = 42)
xgb_model.fit(x_train, y_train)
# predict
y_pred_xgb =xgb_model.predict(x_val)
# calculating the mean squared error
rmse_xgb = np.sqrt(mean_squared_error(y_val, y_pred_xgb))
print("Validation RMSE score: ",round(rmse_xgb,5))
```

###Section 16: Feature Engineering



```python
# now we will be using feature engineering so that the accuracy of the model increses
# creating new cleaned dataframe for modeling
df_fe = df_clean.copy()
# total square footage by combining basement, first and second floor
df_fe['TotalSF'] = df_fe['TotalBsmtSF'] + df_fe['1stFlrSF'] + df_fe['2ndFlrSF']
# total number of bathrooms by combining basemetn fullbath and half bath
df_fe['TotalBath'] = df_fe['BsmtFullBath'] + (0.5 * df_fe['BsmtHalfBath']) + df_fe['FullBath'] + (0.5 * df_fe['HalfBath'])
# house age
df_fe['HouseAge'] = df_fe['YrSold'] - df_fe['YearBuilt']
# remodeled age
df_fe['RemodeledAge'] = df_fe['YrSold'] - df_fe['YearRemodAdd']
# is this house new?
df_fe['IsNew'] = (df_fe['YearBuilt'] == df_fe['YrSold']).astype(int)
print("New features added to the dataframe")
print(df_fe[['TotalSF', 'TotalBath', 'HouseAge', 'RemodeledAge', 'IsNew']].head())
```

###Section 17: Preparing Data After Feature Engineering

```python
# Preparing the data again with the new features
# Making sure LogSalePrice exists
df_fe['LogSalePrice'] = np.log1p(df_fe['SalePrice'])
X_fe = df_fe.drop(['Id', 'SalePrice', 'LogSalePrice'], axis=1)
y_fe = df_fe['LogSalePrice']
# one-hot encode
X_fe_encoded = pd.get_dummies(X_fe)
# Splitting the data again
X_train_fe, X_val_fe, y_train_fe, y_val_fe = train_test_split(X_fe_encoded, y_fe, test_size=0.2, random_state=42)
print("Training set size: ",X_train_fe.shape)
print("Validation set size: ",X_val_fe.shape)
```

### Section 18: Models After Feature Engineering (Data Leakage Detected)

```python
# Linear Regresssion model
lr_model_fe =  LinearRegression()
lr_model_fe.fit(X_train_fe, y_train_fe)
y_pred_lr = lr_model_fe.predict(X_val_fe)
rmse_lr = np.sqrt(mean_squared_error(y_val_fe, y_pred_lr))
# Random Forest model
rf_model_fe = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)
rf_model_fe.fit(X_train_fe, y_train_fe)
y_pred_rf = rf_model_fe.predict(X_val_fe)
rmse_rf = np.sqrt(mean_squared_error(y_val_fe, y_pred_rf))
# XGboost model
xgb_model_fe = XGBRegressor(n_estimators=1000,learning_rate=0.05,max_depth=4,random_state=42,n_jobs=-1)
xgb_model_fe.fit(X_train_fe, y_train_fe)
y_pred_xgb = xgb_model_fe.predict(X_val_fe)
rmse_xgb = np.sqrt(mean_squared_error(y_val_fe, y_pred_xgb))
# result comparision
print("===== After Feature Engineering =====")
print(f"Linear Regression RMSE : {rmse_lr:.5f}")
print(f"Random Forest RMSE     : {rmse_rf:.5f}")
print(f"XGBoost RMSE           : {rmse_xgb:.5f}")
```

### Section 19: Data Leakage Detection
After feature engineering, Linear Regression gave an unrealistic RMSE of 0.00000.  
This shows that there is a possible data leakage.so now I will check whether the target columns were still present in the features.

```python
# Checking if target columns are still in the features as predictions are too accurate
print("Is 'SalePrice' in features?", 'SalePrice' in X_fe_encoded.columns)
print("Is 'LogSalePrice' in features?", 'LogSalePrice' in X_fe_encoded.columns)

# Also checking the first few column names
print("\nFirst 15 columns:")
print(X_fe_encoded.columns[:15].tolist())
```

###Section 20: Confirming the Leakage

```python
# Checking a few actual predictions vs real values
print("First 10 real values (y_val_fe):")
print(y_val_fe.head(10).values)

print("\nFirst 10 predictions (Linear Regression):")
print(y_pred_lr[:10])

print("\nDifference (should not be all zeros):")
print((y_val_fe.values[:10] - y_pred_lr[:10]).round(5))
print("Are we predicting on validation set?")
print("X_val_fe shape:", X_val_fe.shape)
print("y_val_fe shape:", y_val_fe.shape)
```

###Section 21: Finding the Leaking Column
The correlation analysis revealed that a column named `SalePrice_log` had a perfect correlation of 1.0 with the target. This was the source of the data leakage.

```python
# Since this confirms that there is a leakage
# Checking the correlation of every feature with the target
corr_with_target = X_train_fe.corrwith(y_train_fe).abs().sort_values(ascending=False)

print("Top 10 features most correlated with target:")
print(corr_with_target.head(10))
```

###Section 22: Fixing the Data Leakage

```python
# Removing the leaking column
X_fe_encoded = X_fe_encoded.drop(columns=['SalePrice_log'], errors='ignore')

# Spliting the data again
X_train_fe, X_val_fe, y_train_fe, y_val_fe = train_test_split(
    X_fe_encoded,
    y_fe,
    test_size=0.2,
    random_state=42
)

print("Leaking column removed.")
print("New shape:", X_train_fe.shape)
```

###Section 23: Results After Fixing Data Leakage

```python
# Linear Regresssion model
lr_model_fe =  LinearRegression()
lr_model_fe.fit(X_train_fe, y_train_fe)
y_pred_lr = lr_model_fe.predict(X_val_fe)
rmse_lr = np.sqrt(mean_squared_error(y_val_fe, y_pred_lr))
# Random Forest model
rf_model_fe = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)
rf_model_fe.fit(X_train_fe, y_train_fe)
y_pred_rf = rf_model_fe.predict(X_val_fe)
rmse_rf = np.sqrt(mean_squared_error(y_val_fe, y_pred_rf))
# XGboost model
xgb_model_fe = XGBRegressor(n_estimators=1000,learning_rate=0.05,max_depth=4,random_state=42,n_jobs=-1)
xgb_model_fe.fit(X_train_fe, y_train_fe)
y_pred_xgb = xgb_model_fe.predict(X_val_fe)
rmse_xgb = np.sqrt(mean_squared_error(y_val_fe, y_pred_xgb))
# result comparision
print("===== After Feature Engineering =====")
print(f"Linear Regression RMSE : {rmse_lr:.5f}")
print(f"Random Forest RMSE     : {rmse_rf:.5f}")
print(f"XGBoost RMSE           : {rmse_xgb:.5f}")
```

###Section 29: Ordinal Encoding


```python
# defining the ordinal maps
ordinal_mappings = {
    'ExterQual': {'Po': 1, 'Fa': 2, 'TA': 3, 'Gd': 4, 'Ex': 5},
    'ExterCond': {'Po': 1, 'Fa': 2, 'TA': 3, 'Gd': 4, 'Ex': 5},
    'BsmtQual': {'None': 0, 'Po': 1, 'Fa': 2, 'TA': 3, 'Gd': 4, 'Ex': 5},
    'BsmtCond': {'None': 0, 'Po': 1, 'Fa': 2, 'TA': 3, 'Gd': 4, 'Ex': 5},
    'HeatingQC': {'Po': 1, 'Fa': 2, 'TA': 3, 'Gd': 4, 'Ex': 5},
    'KitchenQual': {'Po': 1, 'Fa': 2, 'TA': 3, 'Gd': 4, 'Ex': 5},
    'FireplaceQu': {'None': 0, 'Po': 1, 'Fa': 2, 'TA': 3, 'Gd': 4, 'Ex': 5},
    'GarageQual': {'None': 0, 'Po': 1, 'Fa': 2, 'TA': 3, 'Gd': 4, 'Ex': 5},
    'GarageCond': {'None': 0, 'Po': 1, 'Fa': 2, 'TA': 3, 'Gd': 4, 'Ex': 5},
    'PoolQC': {'None': 0, 'Fa': 1, 'TA': 2, 'Gd': 3, 'Ex': 4}
}
df_ord = df_fe.copy()
# applying the mappings
for col, mapping in ordinal_mappings.items():
    if col in df_ord.columns:
        df_ord[col] = df_ord[col].map(mapping)
# checking the values
print(df_ord[['ExterQual', 'ExterCond', 'BsmtQual', 'BsmtCond', 'HeatingQC', 'KitchenQual', 'FireplaceQu', 'GarageQual', 'GarageCond', 'PoolQC']].head())
```

###Section 30: Preparing Data After Ordinal Encoding

```python
# Preparing the data with ordina; encoding
X_ord = df_ord.drop(['Id', 'SalePrice', 'LogSalePrice'], axis=1)
y_ord = df_ord['LogSalePrice']
# one-hot encoding the remaining categorical colums
X_ord_encoded = pd.get_dummies(X_ord)
# removing any leftover leaking column just in case
X_ord_encoded = X_ord_encoded.drop(columns=['SalePrice_log'], errors='ignore')
#s plitting the data
X_train_ord, X_val_ord, y_train_ord, y_val_ord = train_test_split(X_ord_encoded, y_ord, test_size=0.2, random_state=42)
print("Training set size: ",X_train_ord.shape)
```

###Section 31: Final Model - XGBoost with Ordinal Encoding

```python
# using the xgbord model
xgb_ord = XGBRegressor(n_estimators=1000,learning_rate=0.03,max_depth=4,random_state=42,n_jobs=-1,subsample=0.8,colsample_bytree=0.8,reg_alpha=0.1,reg_lambda = 1.0)
xgb_ord.fit(X_train_ord, y_train_ord)
y_pred_ord = xgb_ord.predict(X_val_ord)
rmse_ord = np.sqrt(mean_squared_error(y_val_ord, y_pred_ord))
print("RMSE with Ordinal Encoding:", round(rmse_ord, 5))
```
